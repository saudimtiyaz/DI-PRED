# Final Results Analysis — Disease Prediction
This notebook compares the performance of **Logistic Regression** and **Random Forest** models across both the **Diabetes** and **Heart Disease** datasets.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

## 1. Load Model Metrics
We load the `results/metrics.csv` file generated by the evaluation script.

```python
df = pd.read_csv('../results/metrics.csv')
df_styled = df.style.highlight_max(axis=0, subset=['Accuracy', 'F1', 'AUC'], color='lightgreen')
display(df_styled)
```

## 2. Visualization
Comparing models side-by-side using bar charts.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Accuracy
sns.barplot(data=df, x='Dataset', y='Accuracy', hue='Model', ax=axes[0], palette='viridis')
axes[0].set_title('Accuracy by Dataset and Model')
axes[0].set_ylim(0, 1.0)
for p in axes[0].patches:
    axes[0].annotate(format(p.get_height(), '.2f'), (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha = 'center', va = 'center', xytext = (0, 9), textcoords = 'offset points')

# AUC
sns.barplot(data=df, x='Dataset', y='AUC', hue='Model', ax=axes[1], palette='magma')
axes[1].set_title('AUC Score by Dataset and Model')
axes[1].set_ylim(0, 1.0)
for p in axes[1].patches:
    axes[1].annotate(format(p.get_height(), '.2f'), (p.get_x() + p.get_width() / 2., p.get_height()), 
                     ha = 'center', va = 'center', xytext = (0, 9), textcoords = 'offset points')

plt.tight_layout()
plt.show()
```

## 3. Final Leaderboard (Best Models)
Based on the analysis, here are the recommended models for each disease prediction task.

```python
best_models = df.sort_values('AUC', ascending=False).drop_duplicates('Dataset')
print("=== BEST MODELS BY DATASET ===")
display(best_models[['Dataset', 'Model', 'Accuracy', 'AUC']])
```

## 4. Observations
1. **Heart Disease:** The **Random Forest** model performed exceptionally well, achieving nearly perfect accuracy (0.98) and AUC (1.00).
2. **Diabetes:** Interestingly, **Logistic Regression** outperformed Random Forest in this dataset, suggesting that a simpler linear boundary might be more effective or that the Random Forest was overfitting the training data.
3. **Next Steps:** Future work should include hyperparameter tuning for the Diabetes Random Forest to see if it can be improved.
